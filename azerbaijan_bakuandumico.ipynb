```python
#go to https://www.whatismybrowser.com/
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/125.0.0.0 Safari/537.36'
}
s = requests.Session()
# r = requests.get("https://www.bakuelectronics.az/catalog/", headers=headers)
# response = BeautifulSoup(r.text, features='html.parser')
```

```python
with s.get("https://www.bakuelectronics.az/catalog/", headers=headers) as r:
    response = BeautifulSoup(r.text, features='html.parser')
```

```python
# categories
categories_list = [
    item.find("a").get("href")
    for item in
    response.find(name="ul", attrs={"class":"nav-list js-main-menu-lvl-1"}).find_all("div", {"class": "main-menu__dropdown_second"})
]
```

```python
categories_list
```

```python
categories = [
    cat.rsplit("/", 2)[-3]
    for cat in categories_list
]
```

```python
# Define the Component of URL

base_url = "https://www.bakuelectronics.az/catalog/"
num_pages = 1

session = requests.Session()
```

```python
# Create lists to store the data
product_categories = []
product_titles = []
product_prices = []
```

```python
# Create a Scraping Loop

### Format of URL: https://www.bakuelectronics.az/catalog/telefonlar-qadcetler/?page=2 ###
for x in categories:
    for page in range(1, num_pages +1 , 1):
        # url = f"{base_url}{x}?page={page}"
        url = f"{x}?page={page}"

        # Send a GET request to the URL
        response = session.get(url, verify=False) #verify should be true and we have to put sertificate on this so theyknow that we are scraping
        print(f"Fetching page: {url}")

        # Check if the request was successful
        if response.status_code == 200:
            
            # Parse the HTML content using BeautifulSoup
            soup = BeautifulSoup(response.content, "html.parser")
            
            # Loop through the product listings and extract the data
            cols = soup.find_all('div', class_='flex__row--start js_pagination_container')
            
            # for row in rows:
            #     cols = row.find_all('div', class_='flex__row--start js_pagination_container')
                
            for col in cols:
                products = col.find_all('div', class_='product__card')
    
                for product in products:
                    category = x
                    title = product.find('a', class_ = 'product__title').get_text()
                    price = product.find('div', class_='product__price--cur').text.strip()
        
                    product_categories.append(category)
                    product_titles.append(title)
                    product_prices.append(price)
    
        time.sleep(3)
        print("sleep 3 seconds")
```

```python
# MPProductsList MPProductsList
import pandas as pd
import requests
import time
from bs4 import BeautifulSoup
from datetime import date
from tqdm import tqdm

!pip install openpyxl
```

```python
#go to https://www.whatismybrowser.com/
nargiz = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/125.0.0.0 Safari/537.36'
}
session = requests.Session()
# r = requests.get("https://www.bakuelectronics.az/catalog/", headers=headers)
# response = BeautifulSoup(r.text, features='html.parser')

url = "https://umico.az/categories"
response = requests.get(url, headers=nargiz)

# Parse the HTML content using BeautifulSoup
soup = BeautifulSoup(response.content, "html.parser")
```

```python
# Find the category listings
katalog = []
categories = soup.find("div", class_ = "MPCategories")
category = categories.find_all("a", href=True)
```

```python
category
```

```python
urlsuperbase = "https://umico.az" 
for cat in category:
    link = urlsuperbase+cat["href"]
    katalog.append(link)

category_list = katalog[3:20] #choose only the categories from dukan to garden
category_list
```

```python
#scraping
n = 1 #how many pages
product_categories = []
product_descs = []
product_prices = []
product_prices_after_discount = [] #if any
product_seller_name = []

for cat in category_list:
    for page in range(1, n+1, 1):
        # url = f"{base_url}{x}?page={page}"
        umico_url = f"{cat}?page={page}"

        # Send a GET request to the URL
        response = session.get(umico_url)
        print(f"Fetching page: {umico_url}")
        
        # Check if the request was successful
        if response.status_code == 200:
            soup_umico = BeautifulSoup(response.content, "html.parser")
            items = soup_umico.find_all("div",{"class": "MPProductItem"})

            for item in items:
                category = cat.rsplit("/", 3)[3]
                # category = cat
                try:
                    price_after_discount = item.find("span",{"data-info": "item-desc-price-new"}).text.strip()
                except Exception as e: 
                    price_after_discount = None
                # print(price_after_discount) #to show the prices

                try:
                    price = item.find("span",{"data-info": "item-desc-price-old"}).text.strip()
                except Exception as e: 
                    price = None

                try:
                    desc = item.find("span",{"class": "MPTitle"}).text.strip()
                except Exception as e: 
                    desc = None
                try:
                    seller = item.find("div",{"class": "line-clamp-2 text-[#1e244d] text-[11px]"}).text.strip()
                except Exception as e: 
                    seller = None

                product_categories.append(category)
                product_descs.append(desc)
                product_prices.append(price)
                product_prices_after_discount.append(price_after_discount)
                product_seller_name.append(seller)
        time.sleep(1) #seconds
print('done')

umico = pd.DataFrame({"Category": product_categories,
                     "Name": product_descs,
                     "Price": product_prices,
                     "Price After Discount": product_prices_after_discount,
                     "Seller": product_seller_name}) 
```
